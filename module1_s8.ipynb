```python
import os
from dotenv import load_dotenv
from mp_api.client import MPRester
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load API Key securely
load_dotenv()
api_key = os.getenv("miP4lnAaROQcFbjLgpZGqM4y6O4NDX4q")
```

```python
# Step 1: Query the API
try:
    with MPRester(miP4lnAaROQcFbjLgpZGqM4y6O4NDX4q) as mpr:
        # Query binary oxides with bandgaps
        docs = mpr.materials.summary.search(
            elements=["O"],
            num_elements=(2, 2),
            has_bandgap=True,
            fields=["material_id", "formula_pretty", "band_gap", "elements"]
        )
# Step 2: Extract to DataFrame
    data_list = []
    for doc in docs:
        # Isolate the dominant element (cation)
        elements_list = [str(el) for el in doc.elements]
        dominant_element = [el for el in elements_list if el != "O"][0]
        
        data_list.append({
            "Material ID": doc.material_id,
            "Formula": doc.formula_pretty,
            "Bandgap (eV)": doc.band_gap,
            "Dominant Element": dominant_element
        })

    df = pd.DataFrame(data_list)
    print(f"✅ Successfully queried {len(df)} binary oxides.\n")
    print(df.head())
# Step 3: Visualize the Distribution
    plt.figure(figsize=(8, 5))
    sns.histplot(df["Bandgap (eV)"], bins=30, kde=True, color="teal")
    plt.title("Bandgap Distribution of Binary Oxides")
    plt.xlabel("Bandgap (eV)")
    plt.ylabel("Frequency")
    plt.show()

except Exception as e:
    print(f"❌ Query Failed: {e}")

```
